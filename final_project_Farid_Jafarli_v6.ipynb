Дерева рішень і градієнтний бустинг до масштабу нечутливі: вони працюють із порядком значень, а не з їхньою величиною, тому для CatBoost нормалізація не потрібна.

Саме тому основною моделлю обрано `CatBoostClassifier`: він обробляє категорії нативно через впорядковані цільові статистики (ordered target statistics), де значення для кожного спостереження рахується лише за попередніми рядками випадкової перестановки, і мітка поточного рядка в нього не потрапляє. CatBoost - алгоритм градієнтного бустингу, тобто ансамблевий метод із теми 9. Це не AutoML-бібліотека: ознаки, модель і параметри задано явно. Порівняння з кодуванням за цільовою змінною наведено в розділі 5.

### Дублювання та узагальнення категорій

Дві ознаки - `Var200` і `Var214` - мають однакову кількість категорій, тому перевіряємо, чи не є вони просто різними позначеннями одного й того ж.

```python
pair = data[['Var200', 'Var214']].dropna()

print(f'рядків без пропусків: {len(pair)}')
print(f'унікальних значень Var200: {pair["Var200"].nunique()}')
print(f'унікальних значень Var214: {pair["Var214"].nunique()}')
print(f'унікальних пар (Var200, Var214): {len(pair.drop_duplicates())}')
```

Кількість унікальних пар збігається з кількістю категорій кожної ознаки, тобто між ними існує взаємно однозначна відповідність: одна ознака повністю дублює іншу.

Додатково дивимось, наскільки категорії взагалі переносяться на валідаційний набір: якщо категорія не зустрічалася під час навчання, модель нічого про неї не знає.

```python
for col in ['Var200', 'Var214', 'Var217', 'Var202']:
    known = data[col].dropna().unique()
    present = test[col].dropna()
    print(f'{col}: категорій {data[col].nunique():>5} | '
          f'пропусків у валідаційній {test[col].isna().mean():>5.1%} | '
          f'нових категорій серед заповнених {(~present.isin(known)).mean():>5.1%}')
```

Пропуски рахуємо окремо від нових категорій, інакше порожні значення валідаційного набору помилково потрапляли б до «нових».

Картина різна. У `Var200` і `Var214` близько 78% заповнених значень валідаційного набору - це категорії, яких не було в навчальних даних, тобто для більшості рядків ці ознаки нічого не підказують. У `Var217` таких значень близько 40%, у `Var202` - 15%.

Тому вилучаємо лише дубльовану пару `Var200` / `Var214`: одна ознака повністю повторює іншу, а переносяться на нові дані вони найгірше. `Var217` і `Var202` залишаємо - перевірка крос-валідацією показала, що їх вилучення навпаки знижує balanced accuracy.

```python
drop_cols = ['Var200', 'Var214']

cols_keep = [col for col in cols_keep if col not in drop_cols]

print(f'колонок для моделі: {len(cols_keep)}')
```

## 3. Підготовка ознак

Обробка мінімальна, оскільки CatBoost самостійно працює з пропусками в числових ознаках і з категоріями. Додаємо одну похідну ознаку - кількість пропусків у рядку: повнота заповнення профілю клієнта сама по собі може бути сигналом.

```python
def prepare(df, cols):
    X = df[cols].copy()
    X['n_missing'] = df.isna().sum(axis=1)
    for col in X.select_dtypes(include='object').columns:
        X[col] = X[col].fillna('NA').astype(str)
    return X


X = prepare(data, cols_keep)
X_test = prepare(test, cols_keep)

cat_cols = X.select_dtypes(include='object').columns.tolist()

print(f'ознак: {X.shape[1]} (категоріальних: {len(cat_cols)})')
```

Для логістичної регресії результат не змінюється (0.8012 проти 0.8009): PCA лише обертає систему координат, а лінійна модель і так працює з лінійними комбінаціями ознак.

Саме ця різниця пояснює механізм. Дерева розбивають простір за окремими ознаками, і кожен поріг має сенс лише у вихідних координатах; після обертання ту саму межу доводиться складати з багатьох косих розбиттів. До того ж стискати тут майже нічого: перша компонента пояснює лише 20% дисперсії, тобто числові ознаки слабко корельовані й надлишковості, яку могло б прибрати PCA, у них немає.

## 5. Основна модель - CatBoost

Дисбаланс класів враховуємо параметром `auto_class_weights='Balanced'`.

Оцінюємо якість 10-кратною стратифікованою крос-валідацією: на кожній ітерації модель навчається лише на тренувальних фолдах, а прогнози для валідаційного фолду зберігаємо (out-of-fold). Так отримуємо оцінку без витоку цільової змінної між фолдами.

Чому саме десять фолдів. Кожна модель при цьому бачить 9000 рядків замість 8000, тобто ближче до фінальної моделі, навченої на всіх 10000, і оцінка виходить менш заниженою: у прямому порівнянні 10 фолдів дають balanced accuracy 0.8994 проти 0.8966 при пʼяти, причому з меншим розкидом між розбиттями (0.0020 проти 0.0033). Десятикратна крос-валідація є усталеним компромісом між зсувом оцінки та обчислювальною вартістю.

Порівняльні експерименти в розділі 4 і перевірки нижче виконувались на пʼяти фолдах - там важливий був відносний порядок варіантів, а не точність абсолютного значення.

```python
PARAMS = dict(iterations=1500,
              learning_rate=0.03,
              depth=6,
              l2_leaf_reg=10,
              auto_class_weights='Balanced',
              random_seed=42,
              verbose=0)


def fit_predict(X_tr, y_tr, X_va):
    model = CatBoostClassifier(**PARAMS)
    model.fit(Pool(X_tr, y_tr, cat_features=cat_cols))
    return model.predict_proba(Pool(X_va, cat_features=cat_cols))[:, 1]
```

```python
cv = StratifiedKFold(n_splits=10, shuffle=True, random_state=42)

proba_cv = np.zeros(len(X))

for train_idx, valid_idx in cv.split(X, target):
    proba_cv[valid_idx] = fit_predict(X.iloc[train_idx],
                                      target.iloc[train_idx],
                                      X.iloc[valid_idx])

print(f'ROC AUC: {roc_auc_score(target, proba_cv):.4f}')
print(f'Balanced accuracy (поріг 0.5): '
      f'{balanced_accuracy_score(target, (proba_cv >= 0.5).astype(int)):.4f}')
```

### Другий учасник - LightGBM

Розділ 4 показав, що ансамбль із RandomForest і HistGradientBoosting програє самотньому CatBoost. Але ті два учасники були слабші за нього на 0.014-0.027, а рівне усереднення нерівних моделей неминуче тягне найсильнішу вниз. Це говорить радше про той конкретний ансамбль, ніж про ансамблі взагалі.

Беремо учасника, який здатен конкурувати: LightGBM теж обробляє категорії нативно. Поодинці він слабший за CatBoost, зате помітно менш схожий на нього - кореляція рангів 0.916 проти 0.955 у XGBoost. Для ансамблю важлива саме несхожість: моделі мають помилятись на різних спостереженнях, інакше усереднювати нічого.

LightGBM позначає категорії внутрішніми кодами, тому навчальний і валідаційний набори мають користуватись спільним словником категорій - інакше фінальна модель прогнозувала б за зміщеними кодами.

```python
LGBM_PARAMS = dict(n_estimators=1500,
                   learning_rate=0.03,
                   num_leaves=31,
                   class_weight='balanced',
                   random_state=42,
                   verbose=-1)

categories = {col: pd.Index(sorted(set(X[col]) | set(X_test[col])))
              for col in cat_cols}


def to_category(frame):
    """Спільний словник категорій для навчальної та валідаційної частин."""
    out = frame.copy()
    for col in cat_cols:
        out[col] = pd.Categorical(out[col], categories=categories[col])
    return out


X_lgbm = to_category(X)
X_test_lgbm = to_category(X_test)

proba_lgbm = np.zeros(len(X))

for train_idx, valid_idx in cv.split(X_lgbm, target):
    model = LGBMClassifier(**LGBM_PARAMS)
    model.fit(X_lgbm.iloc[train_idx], target.iloc[train_idx])
    proba_lgbm[valid_idx] = model.predict_proba(
        X_lgbm.iloc[valid_idx])[:, 1]

print(f'LightGBM ROC AUC: {roc_auc_score(target, proba_lgbm):.4f}')
print(f'кореляція рангів із CatBoost: '
      f'{np.corrcoef(rankdata(proba_cv), rankdata(proba_lgbm))[0, 1]:.3f}')
```

### Суміш

Ймовірності двох моделей мають різний масштаб, тому усереднюємо не самі значення, а ранги - так жодна модель не домінує лише через впевненіші ймовірності.

Вага має бути малою. CatBoost помітно сильніший, і LightGBM потрібен не як рівноправний учасник, а як джерело різноманіття. Нижче будуємо залежність досяжної balanced accuracy від ваги, щоб побачити, де лежить оптимум.

```python
BLEND_WEIGHT = 0.15
RATES = np.arange(0.15, 0.32, 0.001)


def ranks(proba):
    """Ранги в межах набору: важливий лише порядок спостережень."""
    return rankdata(proba) / len(proba)


def blend(proba_a, proba_b, weight=None):
    weight = BLEND_WEIGHT if weight is None else weight
    return (1 - weight) * ranks(proba_a) + weight * ranks(proba_b)


score_cv = blend(proba_cv, proba_lgbm)

print(f'ROC AUC: CatBoost {roc_auc_score(target, proba_cv):.4f} | '
      f'LightGBM {roc_auc_score(target, proba_lgbm):.4f} | '
      f'суміш {roc_auc_score(target, score_cv):.4f}')
```

```python
def rate_to_pred(score, rate):
    """Позначаємо задану частку спостережень із найвищим рангом."""
    pred = np.zeros(len(score), dtype=int)
    pred[np.argsort(score)[::-1][:int(round(rate * len(score)))]] = 1
    return pred


def ba_at_rate(score, y_true, rate):
    return balanced_accuracy_score(y_true, rate_to_pred(score, rate))


def best_ba(score, y_true):
    return max(ba_at_rate(score, y_true, rate) for rate in RATES)


sweep = pd.Series(
    {round(w, 2): best_ba(blend(proba_cv, proba_lgbm, w), target)
     for w in np.arange(0, 0.55, 0.05)})
sweep.index.name = 'вага LightGBM'

ax = sns.lineplot(sweep)
ax.set(ylabel='досяжна balanced accuracy')
ax.axvline(x=BLEND_WEIGHT, c='black', linestyle='--', linewidth=0.75)

plt.show()

sweep.round(4)
```

Крива вийшла пологою: усі значення лежать у смузі приблизно 0.0027, від 0.8999 до 0.9026. Для порівняння, сама оцінка досяжної balanced accuracy має похибку близько 0.002 - це видно з таблиці повторів вище, де вона коливалась від 0.8973 до 0.9012 залежно лише від розбиття. Тобто розкид кривої майже не перевищує власної похибки вимірювання, і вказати оптимум вона не здатна: її максимум цілком може бути шумом.

Тому вагу обираємо не за максимумом кривої, а з міркування про ролі учасників. CatBoost помітно сильніший: ROC AUC 0.9641 проти 0.9598, а за досяжною balanced accuracy розрив іще більший. LightGBM потрібен не як рівноправний учасник, а як джерело незалежного сигналу - він помиляється на інших спостереженнях, і саме це, а не його власна сила, робить суміш кращою. Отже вага має бути малою: беремо 0.15.

Це та сама обережність, що й із операційною точкою нижче. Брати аргмаксимум пологої шумної кривої означає підлаштуватись під конкретне розбиття даних.

### Скільки спостережень позначати

Оскільки суміш - це усереднені ранги, а не ймовірності, говорити про «поріг 0.5» тут немає сенсу: значення показують лише порядок. Рішення формулюється інакше - яку частку клієнтів позначаємо як схильних до відтоку.

Будуємо криву метрики за цією часткою з кроком 0.001, окремо для CatBoost і для суміші.

```python
curves = pd.DataFrame(
    {'CatBoost': [ba_at_rate(ranks(proba_cv), target, r) for r in RATES],
     'суміш': [ba_at_rate(score_cv, target, r) for r in RATES]},
    index=RATES.round(3))

rate_argmax = curves['суміш'].idxmax()

ax = sns.lineplot(curves)
ax.set(xlabel='частка позначених спостережень', ylabel='balanced accuracy')
ax.axvline(x=rate_argmax, c='black', linestyle='--', linewidth=0.75)

plt.show()

print(f'аргмаксимум суміші: частка {rate_argmax:.3f} | '
      f'balanced accuracy {curves["суміш"].max():.4f}')
print(f'для порівняння, CatBoost окремо: {curves["CatBoost"].max():.4f}')
```

### Операційна точка з теорії

Аргмаксимум - поганий оцінювач саме тому, що крива в околі оптимуму полога: мала похибка у значеннях метрики зсуває положення максимуму далеко. Але для balanced accuracy оптимальне правило відоме в замкненому вигляді, і шукати його перебором не потрібно.

Balanced accuracy монотонно повʼязана зі статистикою Юдена `J = TPR - FPR`. Якщо перенести одне спостереження до позитивних, `J` зміниться на `p/π₁ - (1-p)/π₀`, де `p` - ймовірність позитивного класу для цього спостереження, а `π₁`, `π₀` - частки класів. Цей приріст додатний рівно тоді, коли `p·π₀ > (1-p)·π₁`, тобто коли **`p > π₁`**. Отже позначати варто рівно ті спостереження, для яких ймовірність відтоку перевищує апріорну частку класу - 0.1305.

Лишається оцінити цю ймовірність поблизу межі. Класичний шлях - калібрування Платта: підігнати логістичну криву до логітів прогнозів. Але сигмоїда - це припущення про форму, якого ми не перевіряли, і одна двопараметрична крива мусить описати весь діапазон, неминуче жертвуючи точністю десь заради точності деінде.

Оцінюємо ту саму величину без припущень. Упорядковуємо спостереження за спаданням прогнозу, рахуємо частку позитивних у ковзному вікні й беремо точку, де вона перетинає апріорну частку.

Вікно задаємо часткою обсягу - десятою частиною, тобто 1000 спостережень на наших 10000. У такому вікні очікується близько 130 позитивних, що дає відносну похибку оцінки щільності приблизно 9%: достатньо вузько, щоб не розмити межу, і достатньо широко, щоб перетин визначався впевнено. Саме часткою, а не сталим числом: інакше на вибірці вдвічі меншій те саме вікно згладжувало б удвічі сильніше, і перевірка перенесенням нижче міряла б не те правило, яке насправді формує прогноз.

Додаткова перевага: правило спирається лише на **порядок** спостережень, а не на значення. Тому воно застосовується безпосередньо до суміші, тоді як калібруванню потрібні справжні ймовірності, яких усереднення рангів не зберігає. Обидві величини - і кого позначати, і скількох - виводяться з одного обʼєкта.

```python
DENSITY_FRACTION = 10          # вікно - десята частина вибірки


def rate_calibrated(proba, y_true, seed=42):
    """Калібрування Платта, далі поріг = апріорна частка."""
    y_true = np.asarray(y_true)
    clipped = np.clip(proba, 1e-6, 1 - 1e-6)
    logit = np.log(clipped / (1 - clipped)).reshape(-1, 1)
    calibrated = np.zeros(len(proba))

    inner = StratifiedKFold(n_splits=5, shuffle=True, random_state=seed)

    for train_idx, valid_idx in inner.split(logit, y_true):
        model = LogisticRegression().fit(logit[train_idx], y_true[train_idx])
        calibrated[valid_idx] = model.predict_proba(logit[valid_idx])[:, 1]

    return (calibrated >= y_true.mean()).mean()


def local_density(score, y_true, fraction=DENSITY_FRACTION):
    """Частка позитивних у ковзному вікні вздовж упорядкованих прогнозів.

    Вікно задається часткою обсягу, а не сталим числом: інакше правило
    згладжувало б по-різному на різних за розміром вибірках.
    """
    window = max(200, len(score) // fraction)
    order = np.argsort(score)[::-1]
    return (pd.Series(np.asarray(y_true)[order])
            .rolling(window, center=True, min_periods=window // 2)
            .mean())


def rate_density(score, y_true, fraction=DENSITY_FRACTION):
    """Ріжемо там, де локальна щільність падає нижче апріорної частки."""
    density = local_density(score, y_true, fraction).to_numpy()
    below = np.flatnonzero(density < np.mean(y_true))
    return (below[0] if len(below) else len(score)) / len(score)


positive_rate = rate_density(score_cv, target)

print(f'апріорна частка позитивного класу: {target.mean():.4f}')
print(f'аргмаксимум кривої суміші:         {rate_argmax:.4f}')
print(f'калібрування Платта (CatBoost):    {rate_calibrated(proba_cv, target):.4f}')
print(f'перетин щільності:                 {positive_rate:.4f}')
```

### Перевірка перенесенням

Порівнювати варіанти за метрикою на тих самих рядках не можна: аргмаксимум за побудовою максимізує саме її, тож завжди виглядатиме краще. Чесна перевірка - перенесення, тобто рівно та операція, яку ми виконуємо для валідаційного набору: на першій половині виводимо операційну точку, застосовуємо до другої й міряємо там. Друга половина не брала участі у виборі.

Порівнюємо чотири варіанти: CatBoost сам із аргмаксимумом, суміш із аргмаксимумом, суміш із калібруванням Платта і обране правило - суміш із перетином щільності. Плюс «оракул», тобто найкраща можлива частка для другої половини, обрана з підглядуванням у її мітки. Це недосяжна стеля, яка показує, скільки метрики взагалі коштує незнання операційної точки.

Заразом збираємо й самі частки, які кожне правило обирає на видимій половині. Метрика в околі оптимуму полога, тому різні правила можуть давати однакову balanced accuracy при помітно різних частках - і тоді вибирати між ними слід за тим, наскільки стійка їхня оцінка.

Перенавчати нічого не треба - усе рахується з наявних out-of-fold прогнозів.

```python
halves = StratifiedShuffleSplit(n_splits=10, test_size=0.5, random_state=42)


def argmax_rate(score, y_true):
    return RATES[np.argmax([ba_at_rate(score, y_true, r) for r in RATES])]


RULES = {'аргмаксимум': lambda mix_a, cat_a, y_a: argmax_rate(mix_a, y_a),
         'Платт': lambda mix_a, cat_a, y_a: rate_calibrated(cat_a, y_a),
         'щільність': lambda mix_a, cat_a, y_a: rate_density(mix_a, y_a)}

transfer_rows, rate_rows = [], []

for idx_a, idx_b in halves.split(score_cv.reshape(-1, 1), target):
    y_a, y_b = target.iloc[idx_a], target.iloc[idx_b]
    cat_a, cat_b = ranks(proba_cv[idx_a]), ranks(proba_cv[idx_b])
    mix_a, mix_b = ranks(score_cv[idx_a]), ranks(score_cv[idx_b])

    rates = {name: rule(mix_a, proba_cv[idx_a], y_a)
             for name, rule in RULES.items()}
    rate_rows.append(rates)

    transfer_rows.append({
        'CatBoost, аргмаксимум': ba_at_rate(cat_b, y_b, argmax_rate(cat_a, y_a)),
        'суміш, аргмаксимум': ba_at_rate(mix_b, y_b, rates['аргмаксимум']),
        'суміш, Платт': ba_at_rate(mix_b, y_b, rates['Платт']),
        'суміш, щільність': ba_at_rate(mix_b, y_b, rates['щільність']),
        'стеля (оракул)': max(ba_at_rate(mix_b, y_b, r) for r in RATES),
    })

transfer = pd.DataFrame(transfer_rows)
rates = pd.DataFrame(rate_rows)

print('balanced accuracy на невидимій половині')
print(transfer.agg(['mean', 'std']).round(4).to_string())

print('\nчастка, обрана на видимій половині - наскільки вона стійка')
print((rates.agg(['mean', 'std']).round(4)).to_string())
print('\nу перерахунку на рядки валідаційного набору:')
for name in rates:
    print(f'  {name:12} {rates[name].mean() * 2500:.0f} ± {rates[name].std() * 2500:.0f}')
```

| суміш, калібрування Платта | 0.8992 |
| **суміш, перетин щільності** | **0.8996** |

Але вирішальний аргумент у другій таблиці. Аргмаксимум і перетин щільності обирають практично однакову частку - 585 і 583 рядки, - проте розкид у них різний удвічі з половиною. Калібрування Платта стабільніше за всіх (±9), але систематично зміщене: воно обирає 563 рядки проти 583-585 в обох інших правил. Причина - припущення про форму. Перетин щільності такого припущення не робить, тому зміщення не має - і при цьому зберігає більшу частину переваги в стабільності.

3. *Категоріальні ознаки.* Кардинальність сягає 5529 категорій на 10 000 спостережень, тому one-hot кодування непридатне. CatBoost обробляє категорії через впорядковані цільові статистики, LightGBM - через власні розбиття за категоріями. Це не AutoML: ознаки, моделі й параметри задано явно.
4. *Дисбаланс класів.* Частка позитивного класу - 13%. Використано `auto_class_weights='Balanced'` у CatBoost і `class_weight='balanced'` у LightGBM. Самих ваг недостатньо: при стандартному порозі 0.5 balanced accuracy CatBoost становить 0.8884, тому операційну точку розглянуто окремо (пункт 6).
5. *Ансамбль.* Перша спроба (розділ 4) програла, бо рівно усереднювала нерівних учасників. Друга виправляє обидві умови: LightGBM порівнянний за силою (ROC AUC 0.9598 проти 0.9641) і помітно менш схожий на CatBoost - кореляція рангів 0.916. Вагу взято малою (0.15), бо партнер потрібен як джерело незалежного сигналу, а не як рівноправний учасник. Розгортка за вагою вийшла пологою в межах власної похибки, тому вагу обрано з міркування, а не за максимумом кривої.
6. *Операційна точка.* Для balanced accuracy оптимальне правило відоме: позначати спостереження, ймовірність відтоку яких перевищує апріорну частку класу (0.1305). Лишається оцінити цю ймовірність біля межі. Калібрування Платта робить це через сигмоїду - припущення про форму, якого ми не перевіряли, і воно дає систематично нижчу частку (563 рядки проти 583-585 у двох незалежних правил). Тому щільність оцінюємо непараметрично: ковзною часткою позитивних уздовж упорядкування, з вікном у десяту частину вибірки. Отримана частка 0.2321.
7. *Порівняння моделей.* У ML-конвеєрі з однаковою попередньою обробкою: LogisticRegression - 0.8009, RandomForest - 0.8742, HistGradientBoosting - 0.8875. Обидва учасники фінальної суміші випереджають їх усі.
8. *Зменшення розмірності.* PCA (95% дисперсії числових ознак) погіршив HistGradientBoosting із 0.8875 до 0.8405: він застосовний лише до числової частини, а основний сигнал несуть категоріальні ознаки, і лінійні комбінації ускладнюють деревам пошук порогів розбиття.
9. *Фінальний прогноз.* Обидві моделі навчено заново на всіх 10 000 розмічених спостереженнях, їхні прогнози для валідаційного набору усереднено за рангами з тією ж вагою, і до отриманого порядку застосовано частку з крос-валідації. Позитивними позначено 580 із 2500 рядків (23.2%), файл `submission.csv` сформовано у форматі `index,y` і звірено з шаблоном змагання.

**Методологічні зауваження**

- Попередню обробку базових моделей винесено всередину `Pipeline`, тому відновлення пропусків, масштабування та кодування категорій виконуються окремо на кожному фолді й не спричиняють витоку.
- LightGBM позначає категорії внутрішніми кодами, тому навчальний і валідаційний набори користуються спільним словником категорій. Без цього фінальна модель прогнозувала б за зміщеними кодами - помилка, яку крос-валідація не виявила б, бо там усі прогнози будуються в межах одного набору.
- Вікно згладжування задано часткою обсягу, а не сталим числом. Інакше на половині даних воно згладжувало б удвічі сильніше, і перевірка перенесенням міряла б не те правило, яке формує фінальний прогноз.
- Правило спирається лише на порядок спостережень, а не на значення прогнозів. Тому воно застосовується безпосередньо до суміші, тоді як калібруванню потрібні справжні ймовірності, яких усереднення рангів не зберігає.
- Вибір 10 фолдів має і зворотний бік. Оцінка якості менш занижена (0.8994 проти 0.8966 при пʼяти фолдах), але оцінка операційної точки шумніша: σ частки 0.0138 проти 0.0094. Більше фолдів означає сильніше корельовані моделі, і розкид підібраної частки зростає. Це компроміс, а не однозначне покращення.
- Незалежна перевірка на відкладеній вибірці (для CatBoost окремо) дала 0.9046 проти 0.9015 за крос-валідацією, тобто завищення не виявлено - але й спростувати його вона не може: на 2000 спостережень стандартна похибка balanced accuracy становить близько 0.010.
- Результати не відтворюються між платформами побітово: кількість потоків змінює порядок підсумовування з рухомою комою. Той самий ноутбук на різних машинах позначає дещо різну кількість рядків - ще один аргумент за правило зі стійкою оцінкою частки.
- AutoML не використовувався. Усі етапи - EDA, відбір ознак, порівняння моделей, крос-валідація, вибір операційної точки та навчання фінальних моделей - реалізовано явно.

**Що можна спробувати далі**

Перевірено й відкинуто: перебір гіперпараметрів, `boosting_type='Ordered'`, різні значення `max_ctr_complexity`, явні відношення числових ознак, PCA, усереднення за випадковим станом, групування рідкісних категорій, а також суміш із XGBoost - він майже дублює CatBoost (кореляція рангів 0.955) і не додає нічого.

Показово, що ROC AUC суміші становить 0.9656, а balanced accuracy - близько 0.90. Моделі добре впорядковують спостереження за ризиком відтоку, але класи частково перекриваються, і жоден поріг не розділяє їх суттєво краще. Показово й те, що приріст ROC AUC від суміші лише частково переходить у приріст balanced accuracy: AUC усереднює якість упорядкування по всіх порогах, а метрика змагання залежить від околу однієї точки. Це радше межа інформативності наявних ознак, ніж брак налаштування, тому помітного зростання варто очікувати передусім від нових даних про клієнтів.
